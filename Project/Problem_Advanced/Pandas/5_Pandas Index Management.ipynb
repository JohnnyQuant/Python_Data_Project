```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datasets import load_dataset


dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Set Index and Display Rows (2.5.1) - Problem

Set the job_posted_date column as the index of the DataFrame and display the first 5 rows.

```python
df.set_index('job_posted_date', inplace= True)
df.head(5)
```

## 🟩 Reset Index and Drop Existing (2.5.2) - Problem

Reset the index of the DataFrame from the last problem (where job_posted_date column is the index) to the default integer index and drop the existing index. Display the first 5 rows.

```python
df.reset_index(drop = True, inplace=True)
# df.drop(columns='job_posted_date', inplace=True)
```

```python
df.head(5)
```

## 🟩 Sort Index (2.5.3) - Problem

Create a pivot table to calculate the median salary_year_avg for each company_name and sort the index in descending order. Display the first 5 rows.

```python
salary_company = df.pivot_table(values = 'salary_year_avg', index = 'company_name', aggfunc='median')
# pivot_table.sort_index(inplace=True, ascending=False)
# pivot_table.head()
salary_company.sort_index(ascending=False).head(5)
```

## 🟨 Pivot Table Reset & Set Index (2.5.4) - Problem
Create a pivot table to calculate the median salary_year_avg for each job_title_short, reset the index of the resulting pivot table, then set the index to salary_year_avg. Display the first 5 rows.

```python
salary_job_title = df.copy().pivot_table(values = 'salary_year_avg', index = 'job_title_short', aggfunc='median')
salary_job_title.reset_index(inplace=True)
salary_job_title.set_index('salary_year_avg', inplace=True)
salary_job_title
```
